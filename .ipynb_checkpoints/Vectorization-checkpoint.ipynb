### Slicing
Slicing creates an array of indices using a set of three values (`start:stop:step`). A subset of values is also valid. Its use is best explained by example:

```python
a = np.arange(10)
print(f"a        = {a}")

# accessing 5 consecutive elements (start : stop : step)
c = a[2 : 7 : 1]
print("a[2:7:1] = ", c) # prints till stop - 1

# accessing 3 elements by step two

c = a[2 : 7 : 2]
print("a[2:7:2] = ", c)

# access all elements index 3 and above

c = a[3 : ]
print("a[3:]    = ", c)

# access all elements below index 3

c = a[: 3]
print("a[:3]    = ", c)

# access all elements

c = a[:]
print(f"a[:]     = {c}")
```

### Vector Vector element-wise operations
Most of the NumPy arithmetic, logical and comparison operations apply to vectors as well. These operators work on an element-by-element basis. For example 
$$ c_i = a_i + b_i $$

Of course, for this to work correctly, the vectors must be of the same size:

```python
a = np.array([1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])

print(f"Binary operators work element wise: {a + b}")

#try a mismatched vector operation
c = np.array([1, 2])
try:
    d = a + c
except Exception as e:
    print("The error message is:")
    print(e)
```

### Vector Vector dot product
The dot product is a mainstay of Linear Algebra and NumPy. This is an operation used extensively in this course and should be well understood. The dot product is shown below.

Let's implement our own version of the dot product below:

**Using a for loop**, implement a function which returns the dot product of two vectors. The function to return given inputs $a$ and $b$:
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
Assume both `a` and `b` are the same shape.

```python
def own_dot(a, b):
    x = 0
    for i in range (a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {own_dot(a, b)}")
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])

c = np.dot(a, b)

print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

### The Need for Speed: vector vs for loop
We utilized the NumPy  library because it improves speed memory efficiency. Let's demonstrate:

```python
np.random.seed(1)
a = np.random.rand(10000000)  # very large arrays
b = np.random.rand(10000000)

tic = time.time() # start time
c = np.dot(a, b)
toc = time.time() #end time

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capture start time
c = own_dot(a,b)
toc = time.time()  # capture end time

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #remove these big arrays from memory
```

## Matrix Creation
The same functions that created 1-D vectors will create 2-D or n-D arrays. Here are some examples


```python
a = np.zeros((1,5))
print(f"a shape = {a.shape}, a = {a}")   

a = np.zeros((2, 1))                                                                   
print(f"a shape = {a.shape}, a = {a}") 

a = np.random.random_sample((1, 1))  
print(f"a shape = {a.shape}, a = {a}") 

a = np.random.random_sample((5,5))
print(f"a shape = {a.shape}, a = {a}") 
```

```python
# Manual data entry

# NumPy routines which allocate memory and fill with user specified values
a = np.array([[5], [4], [3]]);   print(f" a shape = {a.shape}, np.array: a = {a}")
a = np.array([[5],   # One can also
              [4],   # separate values
              [3]]); #into separate rows
print(f" a shape = {a.shape}, np.array: a = {a}")
```

### Indexing

Matrices include a second index. The two indexes describe [row, column]. Access can either return an element or a row/column. See below:

```python
a = np.arange(6).reshape(-1, 2) # -1 in reshape() tells NumPy to automatically calculate that dimension 
print(f"a.shape: {a.shape}, \na= {a}")

# access an element
print(f"\na[2,0].shape  : {a[2, 0].shape}, a[2, 0] = {a[2, 0]},   type(a[2, 0]) = {type(a[2, 0])} ")

# access a row
print(f"a[2].shape  : {a[2].shape}, a[2] = {a[2]}, type(a[2]) = {type(a[2])} ")
```

**Reshape**  
The previous example used [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) to shape the array.  
`a = np.arange(6).reshape(-1, 2) `   
This line of code first created a *1-D Vector* of six elements. It then reshaped that vector into a *2-D* array using the reshape command. This could have been written:  
`a = np.arange(6).reshape(3, 2) `  
To arrive at the same 3 row, 2 column array.
The -1 argument tells the routine to compute the number of rows given the size of the array and the number of columns.

### Slicing
Slicing creates an array of indices using a set of three values (`start:stop:step`). A subset of values is also valid. Its use is best explained by example:

```python
a = np.arange(20).reshape(-1, 10)
print(f"a = \n {a}")

# access 5 consecutive elements 
print(f"a[0, 2:7:1] = {a[0, 2:7:1]}, a[0, 2:7:1].shape = {a[0, 2:7:1].shape}")

#access 5 consecutive elements (start:stop:step) in two rows
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# access all elements 
print(f"a[:, :] = \n {a[:, :]}, a[:, :].shape = {a[:, :].shape}")


# access all elements in one row (very common usage)
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# same as
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
```
