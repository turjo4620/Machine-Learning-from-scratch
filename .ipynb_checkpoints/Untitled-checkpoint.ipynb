```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('studentPerformance.csv')
print(data)
```

```python
plt.scatter(data.Weekly_Study_Hours, data.Score)
plt.show()
```

```python
def loss_calculation(weight, bias, actual_data):
    total_error = 0
    for i in range(len(actual_data)):
        x = actual_data.iloc[i].Weekly_Study_Hours
        y = actual_data.iloc[i].Score
        total_error +=  (y - (weight * x + bias)) ** 2
    total_error / float(len(actual_data))
```

```python
def gradient_descent(weight_now, bias_now, actual_data, alpha):
    weight_gradient = 0
    bias_gradient = 0

    m = len(actual_data)
    for i in range(m):
        x = actual_data.iloc[i].Weekly_Study_Hours
        y = actual_data.iloc[i].Score

        weight_gradient += (1 / m) * (((weight_now * x + bias_now) - y) * x)
        bias_gradient += (1 / m) * (weight_now * x + bias_now)

    weight = weight_now - weight_gradient * alpha
    bias = bias_now - bias_gradient * alpha
    return weight, bias
```

```python
weight = 0
bias = 0
alpha = 0.001
epochs = 1000
for i in range(epochs):
    if i % 50 == 0:
        print(f"Epoch : {i}")
    weight, bias = gradient_descent(weight, bias, data, alpha)

print(weight, bias)

plt.scatter(data.Weekly_Study_Hours, data.Score, color = 'black')
plt.plot(list(range(0, 30)), [weight * x + bias for x in range(0, 30)], color = 'red')
```
